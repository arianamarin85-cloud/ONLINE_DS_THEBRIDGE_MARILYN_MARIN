## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
#VARIABLES
A = 4
B = "Text"
C = 4.1

#1. Si A y B son equivalentes
print("El resultado de A Y B es: ", A == B)

#2. Si A y C NO son equivalentes
print("El resultado de A Y C es: ", A != C)

#3. Si A es mayor que C
print("El resultado de A Y C es: ", A > C)

#4. Si C es menor o igual que A
print("El resultado de C Y A es: ", C <= A)

#5. Si B NO es equivalente a C
print("El resultado de B Y C es: ", B != C)
````

````python
# En este ejercicio calcularemos la velocidad de una persona en llegar a su trabajo
# Partimos de la siguiente formula para el calculo
#   Velocidad: v = d/t (Distancia recorrida entre el tiempo)

distancia_str = input("Ingresa número de Kilometros desde casa a tu trabajo: ")
distancia_real = distancia_str.replace(",", ".")
distancia_real = float(distancia_real)
print("La distancia ingresada es: ", distancia_real, "Km")

tiempo_str = input("Ingresa el tiempo en llegar desde casa a tu trabajo: ")
tiempo_real = tiempo_str.replace(",", ".")
tiempo_real = float(tiempo_real)
print("El tiempo de demora es: ", tiempo_real, "minutos")

velocidad = distancia_real /  tiempo_real
velocidad = round(velocidad, 2)
print("La velocidad de desplazamiento es: ", velocidad, "Km/minutos")

````
